```python
import pandas as pd
```

```python
df = pd.read_csv('/content/indian_realestate_dataset_1000.csv')
```

```python
df
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
import sqlite3


conn = sqlite3.connect("indian_business.db")

df.to_sql(
    "businesses",
    conn,
    if_exists="replace",
    index=False
)

print("SQL table created:", "businesses")
```

```python
query1 = """ SELECT *FROM businesses; """

display(pd.read_sql_query(query1, conn))
```

```python
query2 = """ SELECT city, property_type FROM businesses; """

display(pd.read_sql_query(query2, conn))
```

```python
query3 = """ SELECT * FROM businesses
WHERE city = 'Bangalore';
"""

display(pd.read_sql_query(query3, conn))
```

```python
query3 = """ SELECT * FROM businesses
WHERE city = 'Bangalore' and price_inr > 1000000;
"""

display(pd.read_sql_query(query3, conn))
```

```python
query4 = """ SELECT * FROM businesses
WHERE city = 'Bangalore' and price_inr >= 5000000;
"""

display(pd.read_sql_query(query4, conn))

```

```python
query5 = """ SELECT * FROM businesses
WHERE property_type != 'Apartment';
"""

display(pd.read_sql_query(query5, conn))
```

```python
query6 = """ SELECT "city", "property_type", "price_inr"
FROM businesses
WHERE  price_inr > 1000000
LIMIT 20 ;
"""

display(pd.read_sql_query(query6, conn))
```

```python
query7 = """
SELECT *
FROM businesses
LIMIT 10;
"""

result = pd.read_sql_query(query7, conn)
display(result)
```

```python
print("Total rows:", pd.read_sql_query(
    "SELECT COUNT(*) AS total_rows FROM businesses;",
    conn
).iloc[0, 0])

print("Total columns:", len(df.columns))

print("First 5 records:")
display(pd.read_sql_query(
    "SELECT * FROM businesses LIMIT 5;",
    conn
))

conn.close()
```
